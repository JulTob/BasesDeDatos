### Parte 1. Desarrollo de la aplicación

* Fecha de venta.
* Número de ticket.
* Artículo.
* Cantidad.
* Precio unitario, expresado en pesetas.

El programa deberá calcular el importe total correspondiente al ticket:
$$\text{Importe Total} = \text{Cantidad} \times \text{Precio Unitario}$$

Finalmente, deberá mostrar los datos introducidos y el importe total calculado.

```python
import os
import csv


def InitializeDatabase(tickets_file):
    # Comprobamos si el archivo NO existe
    if not os.path.exists(tickets_file):
        # Al abrirlo en modo 'w' (write), Python crea el archivo automáticamente
        with open(tickets_file, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            # Creamos la cabecera para identificar las columnas
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'PST', 'PST'])
        print(f"El archivo '{tickets_file}' no existía y ha sido creado con éxito con sus cabeceras.")
    else:
        print(f"El archivo '{tickets_file}' ya existe, no es necesario crearlo de nuevo.")
```

```python
def ValidateTicket(ticket, tickets_file):
    ticket_n = ticket['n_ticket']
    with open(tickets_file, mode='r') as file:
        reader = csv.reader(file)
        next(reader)  # Skip the header row
        next(reader)  # Skip the Units row
        for row in reader:
            if ticket_n == int(row[1]):
                return False
    return True

def RecordTicket(ticket, tickets_file):
    assert ValidateTicket(ticket, tickets_file), "El ticket ya existe"
    with open(tickets_file, mode='a', newline='') as file:
        writer = csv.writer(file)
        writer.writerow([ticket['date'], ticket['n_ticket'], ticket['product'], ticket['quantity'], ticket['price'], ticket['total']])

def PrintData(tickets_file):
    with open(tickets_file, mode='r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
```

### Parte 2. Primer cambio de requisitos: cambio de mano de obra / moneda

La empresa comunica que, a partir de este momento, los precios de los artículos se expresarán en euros en lugar de pesetas, ahora admitiendo un máximo de dos decimales.

Realiza las modificaciones necesarias en el programa desarrollado en la Parte 1. Una vez terminadas las modificaciones:

```python
import os
import csv
from datetime import datetime

# --- Funciones de Lectura y Validación (Estilo de la Parte 1) ---

verify = datetime.strptime

def GetDate():
    try:
        date_txt = input("¿Fecha (DD/MM/AAAA)? ")
        formato = "%d/%m/%Y"
        date = verify(date_txt, formato)
        return date_txt # Mantenemos el string formateado para guardarlo
    except Exception as e:
        raise ValueError("La fecha introducida no tiene un formato válido (DD/MM/AAAA) o no es una fecha real.")

def GetTicketNumber():
    try:
        n_ticket = int(input("¿Número de Ticket? "))
        assert n_ticket > 0, "El número de ticket debe ser un número entero positivo"
        return n_ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del número de ticket: {e}")

def GetArticulo():
    try:
        product = input("¿Artículo? ")
        return product
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del artículo: {e}")

def GetQuantity(product):
    try:
        quantity = int(input(f"¿Cantidad de {product}? "))
        assert quantity > 0, "La cantidad debe ser un entero mayor de 0"
        return quantity
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura de la cantidad: {e}")

def GetPriceEUR():
    try:
        # Ahora leemos un float para admitir decimales en Euros
        price = float(input("¿Precio unitario en Euros? "))
        assert price > 0, "El precio debe ser un valor numérico positivo mayor que 0"

        # Comprobamos que no tenga más de dos decimales
        # Redondeamos a 2 decimales para evitar problemas de precisión float
        price_rounded = round(price, 2)
        if not abs(price - price_rounded) < 1e-9:
            raise ValueError("El precio no puede tener más de dos decimales")

        return price_rounded
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del precio: {e}")

# --- Funciones de Flujo y Formateo ---

def Solicitud_EUR():
    try:
        print("Solicitud de ticket (Euros)")
        print("-------------------------")
        ticket = {}
        ticket['date'] = GetDate()
        ticket['n_ticket'] = GetTicketNumber()
        ticket['product'] = GetArticulo()
        ticket['quantity'] = GetQuantity(ticket['product'])
        ticket['price'] = GetPriceEUR()

        # Importe total en euros redondeado a dos decimales
        ticket['total'] = round(ticket['quantity'] * ticket['price'], 2)
        return ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la solicitud de ticket: {e}")

def Ticket_Print_EUR(ticket):
    print("\nTicket Generado")
    print("---------------")
    print(f"Fecha: {ticket['date']}")
    print(f"Ticket: {ticket['n_ticket']}")
    print(f"Artículo: {ticket['product']}")
    print(f"Cantidad: {ticket['quantity']}")
    print(f"Precio: {ticket['price']:.2f} €")
    print(f"Total: {ticket['total']:.2f} €")

# --- Gestión del Almacenamiento en CSV ---

def InitializeDatabase_EUR(tickets_file):
    # Inicializamos una nueva base de datos para Euros
    if not os.path.exists(tickets_file):
        with open(tickets_file, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'EUR', 'EUR']) # Actualizado a EUR
        print(f"El archivo '{tickets_file}' ha sido creado con éxito para Euros.")
    else:
        print(f"El archivo '{tickets_file}' ya existe.")

def ValidateTicket_EUR(ticket, tickets_file):
    ticket_n = ticket['n_ticket']
    if not os.path.exists(tickets_file):
        return True
    with open(tickets_file, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        try:
            next(reader)  # Cabeceras
            next(reader)  # Tipos de datos
        except StopIteration:
            return True
        for row in reader:
            if row and ticket_n == int(row[1]):
                return False
    return True

def RecordTicket_EUR(ticket, tickets_file):
    assert ValidateTicket_EUR(ticket, tickets_file), f"Error: El ticket Nº {ticket['n_ticket']} ya existe en el fichero."
    with open(tickets_file, mode='a', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow([ticket['date'], ticket['n_ticket'], ticket['product'], ticket['quantity'], ticket['price'], ticket['total']])
    print(f"Ticket Nº {ticket['n_ticket']} guardado correctamente.")



```

```python
def ConvertPesetasToEurosCSV(input_file, output_file):
    """
    Lee un fichero de tickets en pesetas (PST) y genera un nuevo fichero en euros (EUR).
    Aplica la conversión oficial: 1 EUR = 166.386 PTS, redondeando a 2 decimales.
    """
    try:
        # Factor oficial de conversión de Pesetas a Euros
        CONVERSION_FACTOR = 166.386

        if not os.path.exists(input_file):
            raise FileNotFoundError(f"El archivo de origen '{input_file}' no existe.")

        with open(input_file, mode='r', encoding='utf-8') as infile:
            reader = csv.reader(infile)

            # Leemos las cabeceras
            header = next(reader)
            types = next(reader)

            # Modificamos los tipos de las columnas de precio y total a 'EUR'
            new_types = [t if t != 'PST' else 'EUR' for t in types]

            rows_to_write = []
            for row in reader:
                if not row:
                    continue

                # Extraemos los datos manteniendo el formato original
                date = row[0]
                n_ticket = int(row[1])
                product = row[2]
                quantity = int(row[3])

                # Convertimos precio unitario y total de PST (enteros) a EUR (con 2 decimales)
                price_pst = float(row[4])
                total_pst = float(row[5])

                price_eur = round(price_pst / CONVERSION_FACTOR, 2)
                total_eur = round(quantity * price_eur, 2)

                rows_to_write.append([date, n_ticket, product, quantity, price_eur, total_eur])

        # Escribimos el nuevo fichero en Euros
        with open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            writer.writerow(header)
            writer.writerow(new_types)
            writer.writerows(rows_to_write)

        print(f"Migración completada con éxito. Archivo guardado como '{output_file}'.")

    except Exception as e:
        raise ValueError(f"Ocurrió un error durante la conversión de la base de datos: {e}")

# --- Prueba de la función de conversión ---
try:
    # Usamos el fichero 'tickets.csv' generado en la Parte 1
    tickets_original = "tickets.csv"
    tickets_migrated = "tickets_eur.csv"

    print("Iniciando conversión de base de datos de pesetas a euros...")
    ConvertPesetasToEurosCSV(tickets_original, tickets_migrated)

    # Mostramos el contenido para verificar el resultado
    print("\nContenido del nuevo archivo migrado:")
    PrintData(tickets_migrated)
except Exception as e:
    print(f"Error en la prueba: {e}")
```

### Parte 3. Segundo cambio de requisitos: incorporación de un nuevo dato

La empresa decide incorporar a cada ticket un nuevo dato denominado **Descuento**. Dicho descuento:
* Se expresará mediante un porcentaje.
* Deberá estar comprendido entre 0 y 100.

El importe total del ticket deberá calcularse ahora teniendo en cuenta dicho descuento, redondeando el resultado a dos decimales:
$$\text{Importe Total} = \text{Cantidad} \times \text{Precio Unitario} \times \left(1 - \frac{\text{Descuento}}{100}\right)$$

Modifica nuevamente el programa para satisfacer los nuevos requisitos. Una vez realizada la modificación:

```python
import os
import csv
from datetime import datetime

# --- 1. Funciones de Lectura y Validación (Parte 3) ---

verify = datetime.strptime

def GetDiscount():
    try:
        # Solicitamos el porcentaje de descuento
        discount = float(input("¿Descuento (porcentaje de 0 a 100)? "))
        assert 0 <= discount <= 100, "El descuento debe ser un valor porcentual entre 0 y 100"

        # Redondeamos a dos decimales por consistencia de diseño
        return round(discount, 2)
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del descuento: {e}")

# --- 2. Funciones de Flujo, Entrada y Formateo adaptadas ---

def Solicitud_Discount():
    try:
        print("Solicitud de ticket (Con Descuento)")
        print("---------------------------------")
        ticket = {}
        ticket['date'] = GetDate() # Reutiliza GetDate de la Parte 2
        ticket['n_ticket'] = GetTicketNumber() # Reutiliza GetTicketNumber de la Parte 2
        ticket['product'] = GetArticulo() # Reutiliza GetArticulo
        ticket['quantity'] = GetQuantity(ticket['product'])
        ticket['price'] = GetPriceEUR() # Reutiliza GetPriceEUR
        ticket['discount'] = GetDiscount()

        # Cálculo oficial: Total = Cantidad * Precio * (1 - Descuento / 100)
        raw_total = ticket['quantity'] * ticket['price'] * (1 - (ticket['discount'] / 100.0))
        ticket['total'] = round(raw_total, 2)
        return ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la solicitud de ticket: {e}")

def Ticket_Print_Discount(ticket):
    print("\nTicket Generado (Con Descuento)")
    print("------------------------------")
    print(f"Fecha: {ticket['date']}")
    print(f"Ticket: {ticket['n_ticket']}")
    print(f"Artículo: {ticket['product']}")
    print(f"Cantidad: {ticket['quantity']}")
    print(f"Precio: {ticket['price']:.2f} €")
    print(f"Descuento: {ticket['discount']:.2f}%")
    print(f"Total Final: {ticket['total']:.2f} €")
```

```python
# --- 3. Gestión de Almacenamiento en CSV (Parte 3) ---

def InitializeDatabase_Discount(tickets_file):
    if not os.path.exists(tickets_file):
        with open(tickets_file, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Descuento', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'EUR', 'PCT', 'EUR'])
        print(f"El archivo '{tickets_file}' ha sido creado con éxito para Tickets con Descuento.")
    else:
        print(f"El archivo '{tickets_file}' ya existe.")

def ValidateTicket_Discount(ticket, tickets_file):
    ticket_n = ticket['n_ticket']
    if not os.path.exists(tickets_file):
        return True
    with open(tickets_file, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        try:
            next(reader) # Cabecera
            next(reader) # Tipos
        except StopIteration:
            return True
        for row in reader:
            if row and ticket_n == int(row[1]):
                return False
    return True

def RecordTicket_Discount(ticket, tickets_file):
    assert ValidateTicket_Discount(ticket, tickets_file), f"Error: El ticket Nº {ticket['n_ticket']} ya existe en el fichero."
    with open(tickets_file, mode='a', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow([
            ticket['date'],
            ticket['n_ticket'],
            ticket['product'],
            ticket['quantity'],
            ticket['price'],
            ticket['discount'],
            ticket['total']
        ])
    print(f"Ticket Nº {ticket['n_ticket']} con descuento del {ticket['discount']}% guardado correctamente.")
```

```python
# --- 4. Migración de Base de Datos de la Parte 2 a la Parte 3 ---

def ConvertEurosToDiscountCSV(input_file, output_file):
    """
    Importa tickets de la Parte 2 (en Euros sin descuento) a la nueva estructura,
    aplicando por defecto un 0% de descuento a los registros preexistentes.
    """
    try:
        if not os.path.exists(input_file):
            raise FileNotFoundError(f"El archivo de origen '{input_file}' no existe.")

        with open(input_file, mode='r', encoding='utf-8') as infile:
            reader = csv.reader(infile)
            header = next(reader)
            types = next(reader)

            rows_to_write = []
            for row in reader:
                if not row:
                    continue
                date = row[0]
                n_ticket = int(row[1])
                product = row[2]
                quantity = int(row[3])
                price = float(row[4])

                # Asignamos descuento inicial de 0.0%
                discount = 0.0
                total = float(row[5]) # El total sigue siendo el mismo al no haber descuento

                rows_to_write.append([date, n_ticket, product, quantity, price, discount, total])

        # Escribimos en el nuevo formato
        with open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Descuento', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'EUR', 'PCT', 'EUR'])
            writer.writerows(rows_to_write)

        print(f"Migración de Euros a Descuento completada con éxito en '{output_file}'.")
    except Exception as e:
        raise ValueError(f"Ocurrió un error durante la migración de datos: {e}")
```

```python
# --- 5. Bucle de Pruebas de Funcionamiento de la Parte 3 ---

tickets_discount_file = "tickets_discount.csv"
if os.path.exists(tickets_discount_file):
    os.remove(tickets_discount_file)

# Inicializamos el nuevo almacén
InitializeDatabase_Discount(tickets_discount_file)

# Migramos datos previos (de 'tickets_eur.csv' generado en la Parte 2)
try:
    ConvertEurosToDiscountCSV("tickets_eur.csv", tickets_discount_file)
except Exception as e:
    print(f"No se pudo migrar la BD previa (puede no existir aún): {e}")

# Generamos un caso de prueba manual con un 15% de descuento
ticket_desc_test = {
    'date': '16/10/2026',
    'n_ticket': 202,
    'product': 'Calculadora',
    'quantity': 2,
    'price': 12.50,
    'discount': 15.0,
    'total': round(2 * 12.50 * (1 - 15.0 / 100), 2)
}

# Guardamos e imprimimos el ticket de prueba
RecordTicket_Discount(ticket_desc_test, tickets_discount_file)
Ticket_Print_Discount(ticket_desc_test)


# Validamos que impida duplicados de la parte 3
try:
    print("\nIntentando registrar un ticket duplicado en la base de datos de descuentos...")
    RecordTicket_Discount(ticket_desc_test, tickets_discount_file)
except AssertionError as e:
    print(f"Éxito en el control de integridad: {e}")

# Mostramos la base de datos resultante para verificar la estructura
print("\nVisualización final de la Base de Datos con Descuentos:")
PrintData(tickets_discount_file)
```

### Respuestas - Parte 4. Análisis de los resultados

**a) ¿Dónde se encuentra definida en el programa la naturaleza de cada dato utilizado (fecha, número de ticket, cantidad, precio, etc.)?**
> La naturaleza y las restricciones de cada dato se encuentran fuertemente acopladas y definidas directamente en el código de la aplicación. Específicamente:
> * En las **funciones de lectura e inspección** (`GetDate`, `GetTicketNumber`, `GetQuantity`, `GetPriceEUR`, `GetDiscount`), donde se realizan conversiones de tipo explícitas (`int()`, `float()`) y aserciones matemáticas (`assert`).
> * En las **cabeceras meta-informativas del fichero plano** (`tickets_discount.csv`), donde de manera artificial tuvimos que documentar una fila con tipos de datos (`['DATE', 'INT', 'STR', 'INT', 'EUR', 'PCT', 'EUR']`).

**b) ¿Qué partes del programa se han visto afectadas por el cambio de pesetas a euros?**
> El cambio afectó a múltiples niveles debido a la dependencia física de los datos:
> * **Entrada de datos:** Se tuvo que redefinir la función de lectura (`GetPrice`) para admitir números decimales (`float`) con una lógica de validación extra para restringirlos a un máximo de dos decimales.
> * **Estructura lógica interna:** El cálculo de los totales y la visualización de los tickets tuvieron que ser modificados para dar formato a los campos monetarios (`{:.2f} €`).
> * **Persistencia (Bases de datos):** La cabecera del fichero físico CSV tuvo que actualizar los metadatos de tipo de columna pasando de `PST` a `EUR`.
> * **Datos Históricos:** Fue imperativo codificar un script adicional (`ConvertPesetasToEurosCSV`) para transformar y migrar de forma masiva los registros que ya existían previamente en formato antiguo.

**c) ¿Qué partes del programa se han visto afectadas al incorporar el nuevo dato Descuento?**
> Al tratarse de una modificación del esquema (añadir un atributo), el impacto fue estructural:
> * **Añadido de funciones:** Se programó `GetDiscount` desde cero para pedir y validar el porcentaje.
> * **Flujo y lógicas de cálculo:** La función de cálculo (`Solicitud_Discount`) incorporó la fórmula con el factor de descuento, alterando el cálculo matemático del total.
> * **Salida por pantalla:** Se reestructuró la representación visual (`Ticket_Print_Discount`) para visualizar el porcentaje aplicado.
> * **Fichero Físico:** Se cambió el esquema del fichero CSV añadiendo una nueva columna física, lo que obligó a escribir una función de migración para inicializar a un valor neutro (`0.0%`) todos los registros que carecían de ese atributo.

**d) Si los mismos datos fueran utilizados por cinco programas diferentes, ¿qué consecuencias tendrían los cambios anteriores sobre dichos programas?**
> La consecuencia directa sería una **ruptura masiva** por falta de compatibilidad (dependencia física y lógica de los datos). Si los cinco programas leyesen de forma independiente el mismo CSV:
> * Cualquier alteración de columnas (como insertar el Descuento) o el cambio de tipo de datos (de enteros a reales con decimales) haría que el resto de los programas fallasen al intentar leer posiciones del archivo inexistentes o procesar tipos inesperados.
> * Cada uno de los cinco programas tendría que ser reescrito, probado y compilado nuevamente de manera individual para adaptarse al nuevo formato.




















**e) ¿Qué problemas podrían producirse si diferentes programas almacenaran y gestionaran independientemente sus propias copias de los mismos datos?**
> Este enfoque descentralizado da lugar a los problemas clásicos de los sistemas basados en ficheros tradicionales:
> * **Redundancia de datos:** Despilfarro innecesario de almacenamiento al duplicar la misma información.
> * **Inconsistencia de datos:** Si se actualiza el precio de un artículo en el sistema de ventas pero no en el de contabilidad, la información dejaría de coincidir, conviviendo datos contradictorios.
> * **Falta de integridad:** Es muy difícil asegurar que las restricciones de validación (por ejemplo, el formato de fecha o números negativos) se apliquen de igual forma en todos los sistemas.






**f) ¿Qué ocurriría si fuera necesario realizar nuevos cambios en la estructura de los datos en un futuro?**
> Se repetiría el ciclo de ineficiencia: detención de sistemas, reprogramación manual de todas las interfaces que interactúan con el almacenamiento, diseño de scripts complejos de migración de datos históricos y un alto riesgo de introducir errores de consistencia en el ecosistema.


**g) A la vista de la experiencia realizada, ¿qué ventajas consideras que podría tener separar la definición y gestión de los datos de los programas que los utilizan?**
> Separar los datos de las aplicaciones (es decir, utilizar un Sistema de Gestión de Bases de Datos - SGBD) ofrece grandes ventajas:
> * **Independencia de datos (Lógica y Física):** Podemos cambiar la estructura interna de las tablas o el almacenamiento físico de los discos sin necesidad de reescribir ni una sola línea de código en las aplicaciones.
> * **Definición única:** Los metadatos, tipos de columnas y restricciones se definen en el propio motor de base de datos una sola vez.
> * **Seguridad e Integridad centralizadas:** El control de accesos, restricciones de validación y concurrencia quedan unificados en el servidor de la base de datos.

**h) ¿Qué ha sido necesario hacer en el programa para garantizar que no puedan almacenarse dos tickets con el mismo número? ¿Dónde queda definida y comprobada esta restricción?**

> Ha sido necesario programar lógica ad-hoc (`ValidateTicket` / `ValidateTicket_Discount`) que realice una búsqueda secuencial completa leyendo línea por línea el fichero CSV para verificar si el número de ticket introducido ya se encuentra registrado.
Esta restricción de **Clave Primaria** queda definida a nivel de aplicación (lógica procedimental) y es comprobada manualmente antes de la escritura de cada registro en el fichero plano, consumiendo recursos de cómputo que se incrementarán exponencialmente a medida que crezca el número de filas.

### Reflexión final
**¿Sería posible definir los datos una sola vez y permitir que diferentes programas trabajasen con ellos sin tener que incorporar su definición y gestión en cada programa?**

> **Sí, es totalmente posible.** Este es el propósito principal de los **Sistemas de Gestión de Bases de Datos (SGBD)** (como PostgreSQL, MySQL u Oracle).

>Bajo este paradigma, el SGBD actúa como un intermediario entre los datos físicos y los diferentes programas clientes. La estructura, los tipos de datos y las restricciones de integridad (como evitar tickets duplicados mediante *Primary Keys* o restringir valores mediante *Check Constraints*) se definen de forma centralizada **en el esquema de la base de datos** y no en el código fuente de las aplicaciones. Así, cualquier cambio estructural se realiza en un único punto sin romper la compatibilidad de los programas que consumen la información.
