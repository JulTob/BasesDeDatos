### Parte 1. Desarrollo de la aplicación

Desarrolla en Python un programa sencillo que solicite por teclado los siguientes datos correspondientes a un ticket de venta:
* Fecha de venta.
* Número de ticket.
* Artículo.
* Cantidad.
* Precio unitario, expresado en pesetas.

El programa deberá calcular el importe total correspondiente al ticket:
$$\text{Importe Total} = \text{Cantidad} \times \text{Precio Unitario}$$

Finalmente, deberá mostrar los datos introducidos y el importe total calculado.

#### Almacenamiento de los datos
Además de mostrar los datos introducidos, el programa deberá almacenar cada ticket en un fichero denominado `tickets.csv`. Cada nuevo ticket deberá añadirse al fichero conservando la información almacenada anteriormente.

El número de ticket deberá ser único. Antes de almacenar los datos, el programa deberá comprobar la información previamente guardada en el fichero. Si ya existe un ticket con el mismo número, el programa deberá informar de esta circunstancia y no permitirá almacenarlo.

* [Documentación oficial de Python: módulo csv](https://docs.python.org/es/3/library/csv.html)

#### Validación y pruebas
El programa deberá comprobar que:
* La fecha introducida sea válida.
* El número de ticket sea un número entero positivo.
* La cantidad sea un número entero positivo mayor que 0.
* El precio unitario sea un valor numérico positivo entero mayor que 0.

```python
from datetime import datetime
verify = datetime.strptime

def GetDate():
    try:
        date_txt = input("¿Fecha (DD/MM/AAAA)? ")
        formato = "%d/%m/%Y"
        date = verify(date_txt, formato)
        return date
    except Exception as e:
        raise ValueError("La fecha introducida no tiene un formato válido (DD/MM/AAAA) o no es una fecha real.")


def GetTicketNumber():
    try:
        n_ticket = int(input("¿Número de Ticket? "))
        assert n_ticket > 0, "El número de ticket debe ser un número entero positivo"
        return n_ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del número de ticket: {e}")


def GetArticulo():
    try:
        product = input("¿Artículo? ")
        return product
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del artículo: {e}")


def GetQuantity(product):
    try:
        quantity = int(input(f"¿Cantidad de {product}? "))
        assert quantity > 0, "La cantidad debe ser un entero mayor de 0"
        return quantity
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura de la cantidad: {e}")


def GetPrice():
    try:
        price = int(input("¿Precio unitario? "))
        assert price > 0, "El precio debe ser un entero mayor de 0"
        return price
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del precio: {e}")


```

```python
import os
import csv


def InitializeDatabase(tickets_file):
    # Comprobamos si el archivo NO existe
    if not os.path.exists(tickets_file):
        # Al abrirlo en modo 'w' (write), Python crea el archivo automáticamente
        with open(tickets_file, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            # Creamos la cabecera para identificar las columnas
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'PST', 'PST'])
        print(f"El archivo '{tickets_file}' no existía y ha sido creado con éxito con sus cabeceras.")
    else:
        print(f"El archivo '{tickets_file}' ya existe, no es necesario crearlo de nuevo.")
```

```python
def ValidateTicket(ticket, tickets_file):
    ticket_n = ticket['n_ticket']
    with open(tickets_file, mode='r') as file:
        reader = csv.reader(file)
        next(reader)  # Skip the header row
        next(reader)  # Skip the Units row
        for row in reader:
            if ticket_n == int(row[1]):
                return False
    return True

def RecordTicket(ticket, tickets_file):
    assert ValidateTicket(ticket, tickets_file), "El ticket ya existe"
    with open(tickets_file, mode='a', newline='') as file:
        writer = csv.writer(file)
        writer.writerow([ticket['date'], ticket['n_ticket'], ticket['product'], ticket['quantity'], ticket['price'], ticket['total']])

def PrintData(tickets_file):
    with open(tickets_file, mode='r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
```

### Parte 2. Primer cambio de requisitos: cambio de mano de obra / moneda

La empresa comunica que, a partir de este momento, los precios de los artículos se expresarán en euros en lugar de pesetas, ahora admitiendo un máximo de dos decimales.

Realiza las modificaciones necesarias en el programa desarrollado en la Parte 1. Una vez terminadas las modificaciones:

```python
import os
import csv
from datetime import datetime

# --- Funciones de Lectura y Validación (Estilo de la Parte 1) ---

verify = datetime.strptime

def GetDate():
    try:
        date_txt = input("¿Fecha (DD/MM/AAAA)? ")
        formato = "%d/%m/%Y"
        date = verify(date_txt, formato)
        return date_txt # Mantenemos el string formateado para guardarlo
    except Exception as e:
        raise ValueError("La fecha introducida no tiene un formato válido (DD/MM/AAAA) o no es una fecha real.")

def GetTicketNumber():
    try:
        n_ticket = int(input("¿Número de Ticket? "))
        assert n_ticket > 0, "El número de ticket debe ser un número entero positivo"
        return n_ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del número de ticket: {e}")

def GetArticulo():
    try:
        product = input("¿Artículo? ")
        return product
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del artículo: {e}")

def GetQuantity(product):
    try:
        quantity = int(input(f"¿Cantidad de {product}? "))
        assert quantity > 0, "La cantidad debe ser un entero mayor de 0"
        return quantity
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura de la cantidad: {e}")

def GetPriceEUR():
    try:
        # Ahora leemos un float para admitir decimales en Euros
        price = float(input("¿Precio unitario en Euros? "))
        assert price > 0, "El precio debe ser un valor numérico positivo mayor que 0"

        # Comprobamos que no tenga más de dos decimales
        # Redondeamos a 2 decimales para evitar problemas de precisión float
        price_rounded = round(price, 2)
        if not abs(price - price_rounded) < 1e-9:
            raise ValueError("El precio no puede tener más de dos decimales")

        return price_rounded
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la lectura del precio: {e}")

# --- Funciones de Flujo y Formateo ---

def Solicitud_EUR():
    try:
        print("Solicitud de ticket (Euros)")
        print("-------------------------")
        ticket = {}
        ticket['date'] = GetDate()
        ticket['n_ticket'] = GetTicketNumber()
        ticket['product'] = GetArticulo()
        ticket['quantity'] = GetQuantity(ticket['product'])
        ticket['price'] = GetPriceEUR()

        # Importe total en euros redondeado a dos decimales
        ticket['total'] = round(ticket['quantity'] * ticket['price'], 2)
        return ticket
    except Exception as e:
        raise ValueError(f"Ocurrió un error en la solicitud de ticket: {e}")

def Ticket_Print_EUR(ticket):
    print("\nTicket Generado")
    print("---------------")
    print(f"Fecha: {ticket['date']}")
    print(f"Ticket: {ticket['n_ticket']}")
    print(f"Artículo: {ticket['product']}")
    print(f"Cantidad: {ticket['quantity']}")
    print(f"Precio: {ticket['price']:.2f} €")
    print(f"Total: {ticket['total']:.2f} €")

# --- Gestión del Almacenamiento en CSV ---

def InitializeDatabase_EUR(tickets_file):
    # Inicializamos una nueva base de datos para Euros
    if not os.path.exists(tickets_file):
        with open(tickets_file, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            writer.writerow(['Fecha', 'Nº Ticket', 'Artículo', 'Cantidad', 'Precio', 'Total'])
            writer.writerow(['DATE', 'INT', 'STR', 'INT', 'EUR', 'EUR']) # Actualizado a EUR
        print(f"El archivo '{tickets_file}' ha sido creado con éxito para Euros.")
    else:
        print(f"El archivo '{tickets_file}' ya existe.")

def ValidateTicket_EUR(ticket, tickets_file):
    ticket_n = ticket['n_ticket']
    if not os.path.exists(tickets_file):
        return True
    with open(tickets_file, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        try:
            next(reader)  # Cabeceras
            next(reader)  # Tipos de datos
        except StopIteration:
            return True
        for row in reader:
            if row and ticket_n == int(row[1]):
                return False
    return True

def RecordTicket_EUR(ticket, tickets_file):
    assert ValidateTicket_EUR(ticket, tickets_file), f"Error: El ticket Nº {ticket['n_ticket']} ya existe en el fichero."
    with open(tickets_file, mode='a', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow([ticket['date'], ticket['n_ticket'], ticket['product'], ticket['quantity'], ticket['price'], ticket['total']])
    print(f"Ticket Nº {ticket['n_ticket']} guardado correctamente.")



```

```python
def ConvertPesetasToEurosCSV(input_file, output_file):
    """
    Lee un fichero de tickets en pesetas (PST) y genera un nuevo fichero en euros (EUR).
    Aplica la conversión oficial: 1 EUR = 166.386 PTS, redondeando a 2 decimales.
    """
    try:
        # Factor oficial de conversión de Pesetas a Euros
        CONVERSION_FACTOR = 166.386

        if not os.path.exists(input_file):
            raise FileNotFoundError(f"El archivo de origen '{input_file}' no existe.")

        with open(input_file, mode='r', encoding='utf-8') as infile:
            reader = csv.reader(infile)

            # Leemos las cabeceras
            header = next(reader)
            types = next(reader)

            # Modificamos los tipos de las columnas de precio y total a 'EUR'
            new_types = [t if t != 'PST' else 'EUR' for t in types]

            rows_to_write = []
            for row in reader:
                if not row:
                    continue

                # Extraemos los datos manteniendo el formato original
                date = row[0]
                n_ticket = int(row[1])
                product = row[2]
                quantity = int(row[3])

                # Convertimos precio unitario y total de PST (enteros) a EUR (con 2 decimales)
                price_pst = float(row[4])
                total_pst = float(row[5])

                price_eur = round(price_pst / CONVERSION_FACTOR, 2)
                total_eur = round(quantity * price_eur, 2)

                rows_to_write.append([date, n_ticket, product, quantity, price_eur, total_eur])

        # Escribimos el nuevo fichero en Euros
        with open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
            writer = csv.writer(outfile)
            writer.writerow(header)
            writer.writerow(new_types)
            writer.writerows(rows_to_write)

        print(f"Migración completada con éxito. Archivo guardado como '{output_file}'.")

    except Exception as e:
        raise ValueError(f"Ocurrió un error durante la conversión de la base de datos: {e}")

# --- Prueba de la función de conversión ---
try:
    # Usamos el fichero 'tickets.csv' generado en la Parte 1
    tickets_original = "tickets.csv"
    tickets_migrated = "tickets_eur.csv"

    print("Iniciando conversión de base de datos de pesetas a euros...")
    ConvertPesetasToEurosCSV(tickets_original, tickets_migrated)

    # Mostramos el contenido para verificar el resultado
    print("\nContenido del nuevo archivo migrado:")
    PrintData(tickets_migrated)
except Exception as e:
    print(f"Error en la prueba: {e}")
```
